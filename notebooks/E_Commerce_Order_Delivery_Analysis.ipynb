```python
date_columns = [
    "order_purchase_timestamp",
    "order_approved_at",
    "order_delivered_carrier_date",
    "order_delivered_customer_date",
    "order_estimated_delivery_date"
]

for col in date_columns:
    orders[col] = pd.to_datetime(orders[col], errors="coerce")

print("Date conversion completed! ✅")

print("\nUpdated data types:")
print(orders[date_columns].dtypes)
```

```python
# Create time-based features from order timestamps

orders["approval_time_hours"] = (
    orders["order_approved_at"] - orders["order_purchase_timestamp"]
).dt.total_seconds() / 3600

orders["delivery_time_days"] = (
    orders["order_delivered_customer_date"] - orders["order_purchase_timestamp"]
).dt.total_seconds() / 86400

orders["delivery_delay_days"] = (
    orders["order_delivered_customer_date"]
    - orders["order_estimated_delivery_date"]
).dt.total_seconds() / 86400

# Check the new columns
display(
    orders[
        [
            "order_id",
            "order_status",
            "approval_time_hours",
            "delivery_time_days",
            "delivery_delay_days"
        ]
    ].head()
)
```

```python
# Check delivery performance

orders["delivery_status"] = orders["delivery_delay_days"].apply(
    lambda x: "Late" if x > 0 else "On Time/Early"
)

# Count late and on-time/early deliveries
print(orders["delivery_status"].value_counts())

print("\nDelivery Status Percentage:")
print(orders["delivery_status"].value_counts(normalize=True) * 100)
```

```python
# Detect delivery-time outliers using the IQR method

delivered_delivery_times = orders[
    orders["delivery_status"] != "Not Delivered"
]["delivery_time_days"].dropna()

Q1 = delivered_delivery_times.quantile(0.25)
Q3 = delivered_delivery_times.quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

outliers = delivered_delivery_times[
    (delivered_delivery_times < lower_bound) |
    (delivered_delivery_times > upper_bound)
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)
print("Number of outliers:", len(outliers))
print("Outlier Percentage:", round(len(outliers) / len(delivered_delivery_times) * 100, 2), "%")
```
